```python
import kagglehub
import os

path = kagglehub.dataset_download(
    "ameyachopade/housing-price-prediction-dataset"
)

print("Dataset downloaded to:", path)
print("\nFiles available:")

for file in os.listdir(path):
    print(file)
```

```python
import pandas as pd
import os
import glob

csv_files = glob.glob(os.path.join(path, "**", "*.csv"),
                      recursive=True)

print("CSV files found:", csv_files)

for file in csv_files:
    print("\nFile:", os.path.basename(file))
    temp_df = pd.read_csv(file)
    print("Shape:", temp_df.shape)
    print("Columns:", temp_df.columns.tolist())
    display(temp_df.head(5))
```

```python
price_names = {
    "price", "price_inr_lakhs", "house_price",
    "sale_price", "selling_price", "price_in_lakhs",
    "target", "price_inr"
}

candidates = []

for file in csv_files:
    temp = pd.read_csv(file, nrows=5)
    normalized = {
        str(col).strip().lower().replace(" ", "_")
        for col in temp.columns
    }

    if normalized & price_names:
        candidates.append(file)

if not candidates:
    raise ValueError(
        "No recognized price column found. Check Cell 3 "
        "and update price_names to match your dataset."
    )

# Choose the largest matching CSV
data_file = max(candidates, key=os.path.getsize)
df = pd.read_csv(data_file)

print("Selected file:", os.path.basename(data_file))
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python
import numpy as np

# Standardize column names
df.columns = [
    str(col).strip().lower().replace(" ", "_")
    for col in df.columns
]

# Find the target price column
target_candidates = [
    "price", "price_inr_lakhs", "house_price",
    "sale_price", "selling_price", "price_in_lakhs",
    "target", "price_inr"
]

target = next(
    (col for col in target_candidates if col in df.columns),
    None
)

if target is None:
    raise ValueError("Please check the price column name.")

print("Target column:", target)

# Filter Bangalore if a city column exists
city_columns = ["city", "city_name"]
city_col = next(
    (col for col in city_columns if col in df.columns),
    None
)

if city_col:
    print("Cities available:", df[city_col].dropna().unique()[:20])

    df = df[
        df[city_col].astype(str).str.strip().str.lower()
        .isin(["bangalore", "bengaluru"])
    ].copy()

    if df.empty:
        raise ValueError(
            "No Bangalore rows found. Check the city values above."
        )

    print("Bangalore records:", len(df))
else:
    print(
        "WARNING: No city column found. Verify that this dataset "
        "contains Bangalore properties only."
    )

# Convert target price to numeric
df[target] = pd.to_numeric(df[target], errors="coerce")

# Remove missing or invalid prices
df = df.dropna(subset=[target])
df = df[np.isfinite(df[target]) & (df[target] > 0)].copy()

# Remove identifier columns from the features
id_columns = [
    col for col in df.columns
    if col in ["id", "property_id", "index", "unnamed:_0"]
]

X = df.drop(columns=[target] + id_columns)
y = df[target]

# Remove columns that are completely empty
X = X.dropna(axis=1, how="all")

# Remove rows with no usable features
valid_rows = X.notna().any(axis=1)
X = X.loc[valid_rows].copy()
y = y.loc[valid_rows].copy()

# Convert numeric-looking text columns to numbers when appropriate
for col in X.select_dtypes(include="object").columns:
    converted = pd.to_numeric(X[col], errors="coerce")

    if converted.notna().mean() > 0.9:
        X[col] = converted

print("\nFeatures:", X.columns.tolist())
print("Number of records:", len(X))
print("Number of features:", X.shape[1])

display(X.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestRegressor

# Identify numerical and categorical columns
numeric_features = X_train.select_dtypes(
    include=["number", "bool"]
).columns.tolist()

categorical_features = [
    col for col in X_train.columns
    if col not in numeric_features
]

# Prepare numerical features
numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

# Prepare categorical features
categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both preprocessing pipelines
preprocessor = ColumnTransformer([
    ("numeric", numeric_transformer, numeric_features),
    ("categorical", categorical_transformer, categorical_features)
])

# Create the complete model
model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", RandomForestRegressor(
        n_estimators=150,
        random_state=42,
        n_jobs=-1,
        min_samples_leaf=2
    ))
])

# Train the model
model.fit(X_train, y_train)

print("Model training completed successfully!")
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"Mean Absolute Error (MAE): {mae:.3f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.3f}")
print(f"R² Score: {r2:.4f}")
print(f"R² Percentage: {r2 * 100:.2f}%")
```

```python
comparison = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": y_pred
})

comparison["Absolute Error"] = abs(
    comparison["Actual Price"] - comparison["Predicted Price"]
)

display(comparison.head(15))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred, alpha=0.5)

minimum = min(y_test.min(), y_pred.min())
maximum = max(y_test.max(), y_pred.max())

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="-",
    color="red"
)

plt.xlabel("Actual House Price")
plt.ylabel("Predicted House Price")
plt.title("Actual vs Predicted Bangalore House Prices")
plt.tight_layout()
plt.show()
```

```python
# Bangalore House Price Prediction

new_house = {}

print("🏠 Bangalore House Price Prediction")
print("Press Enter to use the default value for any input.\n")

# Helper function to handle empty inputs
def get_input(prompt, default, data_type=str):
    value = input(f"{prompt} (default: {default}): ").strip()

    if value == "":
        return default

    try:
        return data_type(value)
    except ValueError:
        print(f"Invalid input. Using default: {default}")
        return default


# 1. Location
new_house["location"] = get_input(
    "1. Location", "Whitefield"
)

# 2. Total area
new_house["total_sqft"] = get_input(
    "2. Total area (sqft)", 1200.0, float
)

# 3. BHK
new_house["bhk"] = get_input(
    "3. Number of BHK", 2, int
)

# 4. Bathrooms
new_house["bath"] = get_input(
    "4. Number of bathrooms", 2, int
)

# 5. Property type
new_house["property_type"] = get_input(
    "5. Property type", "Apartment"
)

# 6. Furnishing status
new_house["furnishing"] = get_input(
    "6. Furnishing status", "Semi-Furnished"
)

# 7. Parking spaces
new_house["parking"] = get_input(
    "7. Number of parking spaces", 1, int
)


# Prepare input for the trained model
new_house_df = pd.DataFrame([new_house])

# Match the training columns
new_house_df = new_house_df.reindex(columns=X.columns)

# Fill any columns not provided by the 7 inputs
for col in X.columns:
    if col not in new_house:
        if col in numeric_features:
            new_house_df[col] = X_train[col].median()
        else:
            modes = X_train[col].mode(dropna=True)
            new_house_df[col] = (
                modes.iloc[0] if not modes.empty else "Unknown"
            )

# Predict the house price
predicted_price = model.predict(new_house_df)[0]

print("\n🏡 Estimated House Price:", round(predicted_price, 2))
print("Price unit:", target)
```
